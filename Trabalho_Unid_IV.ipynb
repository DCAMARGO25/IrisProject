Você foi contratado para criar um modelo de Machine Learning que classifica espécies de flores Iris
com base em características como comprimento e largura das sépalas e pétalas. Você usará o
TensorFlow para construir, treinar e avaliar o modelo.

```python
# Importar bibliotecas
import tensorflow as tf
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Carregar conjunto de dados Iris
iris = load_iris()
X = iris.data
y = iris.target

# Dividir dados em treino e teste com uma nova semente
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=21)

# Normalizar os dados
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Construir o NOVO modelo
model = tf.keras.models.Sequential([
    tf.keras.layers.Dense(64, activation='relu', input_shape=(4,)),
    tf.keras.layers.Dense(32, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])

# Compilar o modelo
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# Treinar o modelo
model.fit(X_train, y_train, epochs=10, verbose=0)

# Avaliar o modelo
model.evaluate(X_test, y_test)

# Fazer a previsão com a entrada ambígua de versicolor
nova_flor = np.array([[5.1, 3.5, 1.4, 0.2]])
nova_flor_normalizada = scaler.transform(nova_flor)
previsao_bruta = model.predict(nova_flor_normalizada)
previsao_final = np.argmax(previsao_bruta)

# Imprimir o resultado
especies = ['setosa', 'versicolor', 'virginica']
print(f"Probabilidades: {np.round(previsao_bruta, 2)}")
print(f"Previsão do modelo: {especies[previsao_final]}")
# Testar uma Iris setosa
nova_flor_setosa = np.array([[5.1, 3.5, 1.4, 0.2]])
nova_flor_normalizada_setosa = scaler.transform(nova_flor_setosa)
previsao_bruta_setosa = model.predict(nova_flor_normalizada_setosa)
previsao_final_setosa = np.argmax(previsao_bruta_setosa)
print(f"Parâmetros: {nova_flor_setosa}")
print(f"Previsão do modelo: {especies[previsao_final_setosa]}\n")

# Testar uma Iris versicolor
nova_flor_versicolor = np.array([[5.7, 2.8, 4.1, 1.3]])
nova_flor_normalizada_versicolor = scaler.transform(nova_flor_versicolor)
previsao_bruta_versicolor = model.predict(nova_flor_normalizada_versicolor)
previsao_final_versicolor = np.argmax(previsao_bruta_versicolor)
print(f"Parâmetros: {nova_flor_versicolor}")
print(f"Previsão do modelo: {especies[previsao_final_versicolor]}\n")

# Testar uma Iris virginica
nova_flor_virginica = np.array([[6.5, 3.0, 5.2, 2.0]])
nova_flor_normalizada_virginica = scaler.transform(nova_flor_virginica)
previsao_bruta_virginica = model.predict(nova_flor_normalizada_virginica)
previsao_final_virginica = np.argmax(previsao_bruta_virginica)
print(f"Parâmetros: {nova_flor_virginica}")
print(f"Previsão do modelo: {especies[previsao_final_virginica]}\n")
```

```python
# Fazer previsões para o conjunto de teste
previsoes = model.predict(X_test)

# Mapear os rótulos numéricos para os nomes das espécies
especies = ['setosa', 'versicolor', 'virginica']

# Encontrar os erros de previsão
print("--- Flores que foram classificadas incorretamente ---")
for i in range(len(X_test)):
    # Encontra a espécie prevista (índice com a maior probabilidade)
    previsao_modelo = np.argmax(previsoes[i])

    # Encontra a espécie real (rótulo correto)
    especie_real = y_test[i]

    # Se a previsão do modelo for diferente da espécie real, imprima o erro
    if previsao_modelo != especie_real:
        previsao_nome = especies[previsao_modelo]
        real_nome = especies[especie_real]

        print(f"Previsão: {previsao_nome}, Real: {real_nome}")
```
